```python
import pandas as pd
from pathlib import Path
src_path = Path.cwd().parents[1]
src_path
```

```python
# save xlsx as csv
df = pd.read_excel(src_path / 'data/CLEAR/CLEAR_corpus_final.xlsx')
df
```

```python
df[['ID', 'BT_easiness', 'Excerpt']]
```

```python
# replace \n with space
df['Excerpt'] = df['Excerpt'].str.replace('\n', ' ')
df[['ID', 'BT_easiness', 'Excerpt']]
# rename column Excerpt to text, BT_easiness to label
df.rename(columns={'Excerpt': 'text', 'BT_easiness': 'label'}, inplace=True)
# replace double space with single space
df['text'] = df['text'].str.replace('  ', ' ')
df[['ID', 'label', 'text', 'Google WC', 'Sentence Count', 'Paragraphs']]
```

```python
df[['ID', 'label', 'text', 'Google WC', 'Sentence Count', 'Paragraphs']].to_csv(src_path / 'data/CLEAR/CLEAR_corpus_final.csv', index=False)
```
